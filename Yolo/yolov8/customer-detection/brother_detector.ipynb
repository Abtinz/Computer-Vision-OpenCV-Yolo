```python
import cv2
from ultralytics import YOLO
import matplotlib.pyplot as plt
```

```python
model = YOLO('yolov8n.pt')

# Initialize the video capture object - 
cap = cv2.VideoCapture(0)
```

```python
if not cap.isOpened():
    print("Error: Could not open video stream. please check other apps for feasible webcam using")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Perform inference on the frame
    results = model(frame)
    

    # Annotate the frame with detected faces
    annotated_frame = results[0].plot()

    # Display the annotated frame 
    cv2.imshow('YOLOv8 Brothers Detection', annotated_frame)

    # Exit on pressing 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        plt.imshow(results[0].plot()[...,::-1]); plt.title("Brothers in Kai-Caffe-Tehran-Iran")
        break



cap.release()
cv2.destroyAllWindows()
```
