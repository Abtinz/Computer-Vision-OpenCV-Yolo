```python
import cv2
from ultralytics import solutions
```

```python
cap = cv2.VideoCapture("./workout.mp4")
assert cap.isOpened(), "Error reading video file, please check the given url"
```

```python
# Initialize video writer for saving outputs
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
video_writer = cv2.VideoWriter("workouts_detector_output.avi", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
```

```python
#here we initialize gym-detector with its parameters ...
gym = solutions.AIGym(
    show=True,
    kpts=[6, 8, 10],
    model="yolo11n-pose.pt"
)
```

```python
while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        print("Video frame is empty or processing is complete.")
        break

    results = gym(frame)
    #Log results for monitorizing
    print(results)
    # write the processed frame with detectedd positions and completed tasks by yolov11 over current frame.
    video_writer.write(results.plot_im)

cap.release()
video_writer.release()
cv2.destroyAllWindows()
```
